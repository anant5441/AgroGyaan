```python
# Example: mapping numeric classes to crop names
crop_labels = {
    0:"apple",
    1:"banana",
    2:"blackgram",
    3:"chickpea",
    4:"coconut",
    5:"coffee",
    6:"cotton",
    7:"grapes",
    8:"jute",
    9:"kidneybeans",
    10:"lentil",
    11:"maize",
    12:"mango",
    13:"mothbeans",
    14:"mungbean",
    15:"muskmelon",
    16:"orange",
    17:"papaya",
    18:"pigeonpeas",
    19:"pomegranate",
    20:"rice",
    21:"watermelon"
}
```

```python
import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
XGBoost=joblib.load('xgb_crop_model.pkl')
best_params = joblib.load('best_params_xgb.pkl')
feature_names = joblib.load('feature_names.pkl')
```

```python
test_samples = [
    [120, 55, 30, 25.4, 70.2, 6.2, 85.0],    # maize
    [85, 58, 60, 27.5, 65.0, 7.0, 60.5],     # cotton
    [70, 50, 40, 28.0, 70.0, 6.7, 120.0],    # mango 
]
```

```python
import numpy as np

# Assume test_samples is a list or array of shape (num_samples, num_features)
for i, sample in enumerate(test_samples):
    
    # Reshape sample for scikit-learn (1 sample, 7 features)
    sample = np.array(sample).reshape(1, -1)
    
    # Prediction
    prediction = XGBoost.predict(sample)[0]
    
    # Probabilities for all classes
    probabilities = XGBoost.predict_proba(sample)[0]
    
    # Get predicted crop name and confidence
    predicted_crop = crop_labels[prediction].replace('crops_', '')
    confidence = probabilities[list(crop_labels.keys()).index(prediction)]
    
    print(f"Sample {i+1}: {sample.flatten()}")
    print(f"Predicted crop: {predicted_crop} (confidence: {confidence:.3f})")
    
    # Top 3 predictions
    top_3_indices = np.argsort(probabilities)[-3:][::-1]
    print("Top 3 predictions:")
    for idx in top_3_indices:
        # Map index to actual class label
        class_label = list(crop_labels.keys())[idx]
        crop_name = crop_labels[class_label].replace('crops_', '')
        print(f"  - {crop_name}: {probabilities[idx]:.3f}")
    print("-" * 30)
```
