```python
import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
model = joblib.load('crop_prediction_model.pkl')
scaler = joblib.load('scaler.pkl')
feature_names = joblib.load('feature_names.pkl')
crop_labels = joblib.load('crop_labels.pkl')
```

```python
print("✅ Model and artifacts loaded successfully!")
print("Feature names:", feature_names)
print("Crop labels:", [label.replace('crops_', '') for label in crop_labels])
print("Number of crop classes:", len(crop_labels))
```

# Test 1

```python
test_samples = [
    [90, 42, 43, 20.88, 82.0, 6.5, 202.94],
    [40, 40, 40, 25.0, 80.0, 7.0, 150.0],    
    [80, 50, 50, 30.0, 60.0, 6.8, 100.0]     
]
```

```python
for i, sample in enumerate(test_samples):
    
    scaled_sample = scaler.transform([sample])
    
    prediction = model.predict(scaled_sample)
    probabilities = model.predict_proba(scaled_sample)[0]
    
    predicted_crop = crop_labels[prediction[0]].replace('crops_', '')
    confidence = probabilities[prediction[0]]
    
    print(f"Sample {i+1}: {sample}")
    print(f"Predicted crop: {predicted_crop} (confidence: {confidence:.3f})")
    
    top_3_indices = np.argsort(probabilities)[-3:][::-1]
    print("Top 3 predictions:")
    for idx in top_3_indices:
        crop_name = crop_labels[idx].replace('crops_', '')
        print(f"  - {crop_name}: {probabilities[idx]:.3f}")
    print("-" * 30)
```

# Test 2

```python
print("✅ Final Validation:")
print("-" * 50)

sample = {
    "N": 90,
    "P": 42, 
    "K": 43,
    "temperature": 20.88,
    "humidity": 82.0,
    "ph": 6.5,
    "rainfall": 202.94
}

input_array = np.array([[sample[feature] for feature in feature_names]])
scaled_input = scaler.transform(input_array)
prediction = model.predict(scaled_input)[0]
confidence = model.predict_proba(scaled_input)[0][prediction]

print(f"Input: {sample}")
print(f"Predicted: {crop_labels[prediction].replace('crops_', '')}")
print(f"Confidence: {confidence:.3f}")
print("🎉 Model is ready for deployment!")
```
